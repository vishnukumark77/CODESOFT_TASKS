##TASK 01- CHATBOT WITH RULE-BASED


```python


print("===============================================")
print("       Welcome to Simple  Chatbot - Effusive")
print("===============================================")


while True:
    user_input = input("You: ").lower().strip()

    # Rule 1: Greeting
    if user_input in ["hello", "hi", "hey", "hii"]:
        print("Bot: Hello! How can I help you?")

    # Rule 2: Asking name
    elif "your name" in user_input or "who are you" in user_input:
        print("Bot: I am a simple rule-based chatbot.")

    # Rule 3: Asking how the bot is
    elif "how are you" in user_input:
        print("Bot: I am fine. Thank you for asking!")

    # Rule 4: Asking about capabilities
    elif "what can you do" in user_input:
        print("Bot: I can answer simple questions based on predefined rules.")

    # Rule 5: Asking for help
    elif "help" in user_input:
        print("Bot: Sure! You can ask me about my name, capabilities, or say hello.")

    # Rule 6: Thank you
    elif "thank" in user_input:
        print("Bot: You're welcome!")

    elif "effu" in user_input:
        print("Bot: Yes! Ask me any thing!")

    elif "what you do" in user_input:
        print("Bot: I'm your vertual friend - You can call me  effu , ask me any thing!")

    elif "vertual internship" in user_input:
        print("Bot: Oh great! You are looking for a best tech company for your Intership ... I have one good option for you ,check for CodSoft - There you will get real industry oriented experience!")

    elif "about codsoft" in user_input:
        print("Bot: CodSoft is an innovative tech education and software development platform designed to empower aspiring developers. Our goal is to transform theoretical knowledge into practical expertise- by CodSoft!")

    # Rule 7: Goodbye
    elif user_input in ["bye", "goodbye", "exit", "quit"]:
        print("Bot: Goodbye! Have a nice day.")
        break

    # Rule 8: Unknown input
    else:
        print("Bot: Sorry, I don't understand that.")
```
